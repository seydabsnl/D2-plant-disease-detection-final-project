```python
import os
import shutil
import random
from pathlib import Path
from sklearn.model_selection import train_test_split

# --- AYARLAR ---
SOURCE_DIR = "./PlantDoc-Original" 
DEST_DIR = "./PlantDoc-Split"       
SEED = 42

def smart_split_dataset():
    
    if os.path.exists(DEST_DIR):
        print(f"Eski {DEST_DIR} siliniyor...")
        shutil.rmtree(DEST_DIR)
    
    # Resim uzantıları
    valid_extensions = {'.jpg', '.jpeg', '.png', '.bmp'}
    
    
    class_images = {}
    
    print("Dosyalar taranıyor...")
    
    
    for root, dirs, files in os.walk(SOURCE_DIR):
        for file in files:
            file_path = Path(root) / file
            if file_path.suffix.lower() in valid_extensions:
                
                class_name = os.path.basename(root)
                
                
                if class_name.lower() in ['train', 'test', 'val', 'validation']:
                    continue
                
                if class_name not in class_images:
                    class_images[class_name] = []
                class_images[class_name].append(str(file_path))

    print(f"Toplam {len(class_images)} farklı sınıf bulundu.")

    
    for class_name, images in class_images.items():
        if len(images) < 5:
            print(f"UYARI: '{class_name}' sınıfında yeterli resim yok ({len(images)}), atlanıyor.")
            continue

        print(f"İşleniyor: {class_name} ({len(images)} resim)")
        
        # 1. Split: Train (%80) ve Geri Kalan (%20)
        try:
            train_imgs, temp_imgs = train_test_split(images, test_size=0.2, random_state=SEED)
            val_imgs, test_imgs = train_test_split(temp_imgs, test_size=0.5, random_state=SEED)
        except ValueError:
            print(f"HATA: '{class_name}' bölünemedi. Resim sayısı çok az olabilir.")
            continue
        
        # Dosyaları kopyala
        splits = {'train': train_imgs, 'val': val_imgs, 'test': test_imgs}
        
        for split_name, split_imgs in splits.items():
            save_dir = os.path.join(DEST_DIR, split_name, class_name)
            os.makedirs(save_dir, exist_ok=True)
            
            for img_path in split_imgs:
                file_name = os.path.basename(img_path)
                dst_path = os.path.join(save_dir, file_name)
                
                if os.path.exists(dst_path):
                    base, ext = os.path.splitext(file_name)
                    dst_path = os.path.join(save_dir, f"{base}_copy{ext}")
                
                shutil.copy(img_path, dst_path)

    print(f"\n✅ İşlem Tamamlandı! Veriler şuraya hazırlandı: {DEST_DIR}")

smart_split_dataset()
```
