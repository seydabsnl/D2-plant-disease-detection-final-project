```python
from ultralytics import YOLO
import os


data_path = os.path.abspath("./PlantDoc-Split")


model = YOLO('yolov8n-cls.pt') 

print("YOLOv8 Eğitimi Başlıyor...")


results = model.train(
    data=data_path, 
    epochs=10, 
    imgsz=224, 
    batch=16,
    project="PlantDoc_YOLO_Project",
    name="yolo_cls_run"
)


print("\nEğitim Tamamlandı!")
print(f"Val Accuracy (Top1): {results.box.map if hasattr(results, 'box') else 'Metrikler klasörde'}")

```
